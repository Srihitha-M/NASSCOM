```python
import pandas as pd
import numpy as np

print("--- Intermediate Pandas Exercise ---")
print("Complete each task by writing the requested Pandas code.")
print("------------------------------------")
```

```python
print("\n--- Task 1: Basic Pandas Functions and Converting Arrays to DataFrames ---")

# 1.1 Create a Pandas Series named 's1_1' from a list of numbers [10, 20, 30, 40, 50]
s1_1 = pd.Series([10, 20, 30, 40, 50])

# 1.2 Create a 3x4 NumPy array named 'np_array1_2' with random integers between 1 and 100
np.random.seed(42)  # For reproducible results
np_array1_2 = np.random.randint(1, 101, size=(3, 4))

# 1.3 Convert 'np_array1_2' into a Pandas DataFrame named 'df1_3' with columns 'A', 'B', 'C', 'D'
df1_3 = pd.DataFrame(np_array1_2, columns=['A', 'B', 'C', 'D'])

print("\ns1_1:")
print(s1_1)
print("\ndf1_3:")
print(df1_3)
```

```python
print("\n--- Task 2: Synthetic Data Generation for Practice ---")

np.random.seed(42)

# Generate 100 rows of synthetic sales data
dates = pd.date_range(start='2023-01-01', end='2023-04-09', periods=100)
regions = np.random.choice(['East', 'West', 'North', 'South'], size=100)
products = np.random.choice(['Laptop', 'Mouse', 'Keyboard', 'Monitor'], size=100)
sales = np.random.randint(100, 1001, size=100).astype(float)
quantity = np.random.randint(1, 11, size=100).astype(float)

df_sales = pd.DataFrame({
    'Date': dates,
    'Region': regions,
    'Product': products,
    'Sales': sales,
    'Quantity': quantity
})

# Introduce 5 random NaN values in 'Sales' and 3 in 'Quantity'
sales_nan_idx = np.random.choice(100, size=5, replace=False)
quantity_nan_idx = np.random.choice(100, size=3, replace=False)

df_sales.loc[sales_nan_idx, 'Sales'] = np.nan
df_sales.loc[quantity_nan_idx, 'Quantity'] = np.nan

print("df_sales head:")
print(df_sales.head())
```

```python
print("\n--- Task 3: Indexing and Slicing in DataFrames ---")

# 3.1 Select only the 'Region' and 'Sales' columns
df3_1 = df_sales[['Region', 'Sales']]

# 3.2 Select rows where 'Product' is 'Laptop'
df3_2 = df_sales[df_sales['Product'] == 'Laptop']

# 3.3 Select rows where 'Region' is 'East' AND 'Sales' are greater than 500
df3_3 = df_sales[(df_sales['Region'] == 'East') & (df_sales['Sales'] > 500)]

# 3.4 Select the 'Sales' value for the first entry where 'Product' is 'Monitor'
monitor_df = df_sales[df_sales['Product'] == 'Monitor']
first_monitor_idx = monitor_df.index[0]
val3_4 = df_sales.loc[first_monitor_idx, 'Sales']

print("val3_4 (First Monitor Sales):", val3_4)
```

```python
print("\n--- Task 4: Data Cleaning (Handling Missing Values) ---")

# 4.1 Check how many missing values are in each column
missing_counts4_1 = df_sales.isnull().sum()

# 4.2 Fill missing 'Sales' values with the mean of the 'Sales' column
df4_2_filled = df_sales.copy()
df4_2_filled['Sales'] = df4_2_filled['Sales'].fillna(df4_2_filled['Sales'].mean())

# 4.3 Drop rows that have any missing values in 'df_sales'
df4_3_dropped = df_sales.dropna()

print("missing_counts4_1:")
print(missing_counts4_1)
```

```python
print("\n--- Task 5: Data Manipulation (Columns, Sorting, Grouping) ---")

# 5.1 Create a new column 'Total_Revenue' which is 'Sales' * 'Quantity'
df_sales['Total_Revenue'] = df_sales['Sales'] * df_sales['Quantity']

# 5.2 Sort 'df_sales' by 'Date' ascending, then by 'Sales' descending
df5_2_sorted = df_sales.sort_values(by=['Date', 'Sales'], ascending=[True, False])

# 5.3 Group 'df_sales' by 'Region' and calculate sum of 'Sales' and 'Quantity'
df5_3_grouped_region = df_sales.groupby('Region')[['Sales', 'Quantity']].sum()

# 5.4 Group 'df_sales' by 'Product' and find average 'Sales'
df5_4_grouped_product = df_sales.groupby('Product')['Sales'].mean()

print("\n--- 5.3 Grouped by Region (Sum) ---")
print(df5_3_grouped_region)

print("\n--- 5.4 Grouped by Product (Mean Sales) ---")
print(df5_4_grouped_product)
```
