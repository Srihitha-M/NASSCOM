```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Sample data
data = ['Red', 'Green', 'Blue', 'Red']

# Initialize and transform
le = LabelEncoder()
encoded = le.fit_transform(data)

# Print raw array and mapped categories
print("Encoded array:", encoded)
mapping = dict(zip(le.classes_, le.transform(le.classes_)))
print("Category Mapping:", mapping)
```

```python
import numpy as np
from sklearn.preprocessing import OneHotEncoder

# Sample data (reshaped to 2D array)
data = np.array([['Red'], ['Green'], ['Blue'], ['Red']])

# Initialize and fit-transform
ohe = OneHotEncoder(sparse_output=False)  # sparse_output=False returns a dense array directly
encoded_matrix = ohe.fit_transform(data)

# Display result alongside category column names
categories = ohe.categories_[0]
encoded_df = pd.DataFrame(encoded_matrix, columns=categories)
print(encoded_df)
```

```python
import pandas as pd

# Sample dataframe
df = pd.DataFrame({'Color': ['Red', 'Green', 'Blue', 'Red']})

# Generate one-hot encoded columns (returns 1s and 0s)
encoded_df = pd.get_dummies(df, columns=['Color'], dtype=int)
print(encoded_df)
```
