```python
# ==============================================================================
# U7 — DS Lifecycle & Data Sources: Lab (Complete Single-Cell Version)
# ==============================================================================

import pandas as pd
import json
import requests
from collections import Counter

print('--- SETUP ---')
print('Setup complete. pandas', pd.__version__, '\n')


# ------------------------------------------------------------------------------
# 1. THE DATA SCIENCE LIFECYCLE
# ------------------------------------------------------------------------------
print('--- 1. THE DATA SCIENCE LIFECYCLE ---')

lifecycle = [
    'Define',    # frame the business problem
    'Collect',   # gather & ingest data
    'Prepare',   # clean & transform
    'Explore',   # EDA & visualise
    'Model',     # train algorithms
    'Evaluate',  # test & validate
    'Deploy',    # serve predictions
    'Monitor',   # watch & maintain  (then loop back!)
]

what_happens = {
    'Define':   'turn a business question into an ML problem',
    'Collect':  'pull data from files, databases, APIs',
    'Prepare':  'clean missing values, fix types, engineer features',
    'Explore':  'summarise & visualise to find patterns',
    'Model':    'train and tune algorithms',
    'Evaluate': 'measure performance on unseen data',
    'Deploy':   'put the model into production',
    'Monitor':  'track performance and retrain as needed',
}

# 🧪 LAB EXERCISE 1 — Order the lifecycle
shuffled = ['Model', 'Define', 'Deploy', 'Collect', 'Monitor',
            'Prepare', 'Evaluate', 'Explore']

# 1 & 2. sort 'shuffled' by position in 'lifecycle'
corrected = sorted(shuffled, key=lifecycle.index)

print('\nLab Ex 1 Results:')
print('  Corrected Order:')
for i, stage in enumerate(corrected, start=1):
    print(f'    {i}. {stage}')

# 3. confirm it matches the reference order
print('  Matches lifecycle reference order?', corrected == lifecycle, '\n')


# ------------------------------------------------------------------------------
# 2. CRISP-DM
# ------------------------------------------------------------------------------
print('--- 2. CRISP-DM ---')

crisp_dm = [
    'Business Understanding',
    'Data Understanding',
    'Data Preparation',
    'Modeling',
    'Evaluation',
    'Deployment',
]

mapping = {
    'Define':   'Business Understanding',
    'Collect':  'Data Understanding',
    'Explore':  'Data Understanding',
    'Prepare':  'Data Preparation',
    'Model':    'Modeling',
    'Evaluate': 'Evaluation',
    'Deploy':   'Deployment',
    'Monitor':  'Deployment',
}

# 🧪 LAB EXERCISE 2 — CRISP-DM grouping
# 1. invert the mapping: group stages by phase
grouped = {}
for stage, phase in mapping.items():
    grouped.setdefault(phase, []).append(stage)

print('\nLab Ex 2 Results:')
print('  Phase -> Stages mapping:')
for phase, stages in grouped.items():
    print(f'    {phase:22s} -> {stages}')

# 3. which phase has the most stages?
most_stages_phase = max(grouped, key=lambda p: len(grouped[p]))
print(f'  Phase covering most stages: {most_stages_phase} ({len(grouped[most_stages_phase])} stages)\n')


# ------------------------------------------------------------------------------
# 3. CLASSIFYING DATA TYPES
# ------------------------------------------------------------------------------
print('--- 3. CLASSIFYING DATA TYPES ---')

def classify(fmt):
    fmt = fmt.lower()
    if fmt in ('csv', 'sql', 'parquet', 'excel', 'xlsx'):
        return 'structured'
    if fmt in ('json', 'xml', 'yaml', 'log'):
        return 'semi-structured'
    return 'unstructured'   # text, image, audio, video, ...

# 🧪 LAB EXERCISE 3 — Classify these datasets
to_label = ['parquet', 'yaml', 'png', 'csv', 'wav', 'json', 'txt']

# 1. classify each one
classified = [classify(fmt) for fmt in to_label]

print('\nLab Ex 3 Results:')
for fmt, cat in zip(to_label, classified):
    print(f'  {fmt:8s} -> {cat}')

# 2. count per category
counts = Counter(classified)
print('  Category Counts:', dict(counts))

# 3. An ML task that uses unstructured data:
# Image classification (e.g., detecting objects using CNNs) or text sentiment analysis.
print('  Unstructured ML Task Example: Image classification on PNG/JPEG images.\n')


# ------------------------------------------------------------------------------
# 4. LOADING DATA FROM FILES (CSV & JSON)
# ------------------------------------------------------------------------------
print('--- 4. LOADING DATA FROM FILES (CSV & JSON) ---')

# Demo 4A/4B setup
sample = pd.DataFrame({
    'name':   ['Ana', 'Bo', 'Cy', 'Di'],
    'age':    [30, 25, 41, 38],
    'city':   ['Pune', 'Delhi', 'Pune', 'Hyderabad'],
    'spend':  [120.5, 80.0, 200.2, 150.0],
})
sample.to_csv('people.csv', index=False)
df = pd.read_csv('people.csv')

# Demo 4C setup
records = [
    {'id': 1, 'user': {'name': 'Ana', 'city': 'Pune'}, 'spend': 120.5},
    {'id': 2, 'user': {'name': 'Bo',  'city': 'Delhi'}, 'spend': 80.0},
]
json_df = pd.json_normalize(records)

# 🧪 LAB EXERCISE 4 — Round-trip a CSV
# 1. build the DataFrame
products_df = pd.DataFrame({
    'product':  ['Laptop', 'Mouse', 'Keyboard'],
    'price':    [999.99, 25.50, 75.00],
    'in_stock': [True, True, False]
})

# 2. write to 'products.csv' then read it back
products_df.to_csv('products.csv', index=False)
loaded_products_df = pd.read_csv('products.csv')

# 3. inspect shape & dtypes
print('\nLab Ex 4 Results:')
print('  Loaded Products DataFrame:')
print(loaded_products_df)
print('  Shape:', loaded_products_df.shape)
print('  Data Types:\n', loaded_products_df.dtypes, '\n')


# ------------------------------------------------------------------------------
# 5. LOADING DATA FROM A REST API
# ------------------------------------------------------------------------------
print('--- 5. LOADING DATA FROM A REST API ---')

# 🧪 LAB EXERCISE 5 — Pull from an API
POSTS_URL = 'https://jsonplaceholder.typicode.com/posts'

# 1. fetch (try/except with a small inline fallback)
try:
    resp = requests.get(POSTS_URL, timeout=10)
    resp.raise_for_status()
    posts_data = resp.json()
    print('  Fetched', len(posts_data), 'posts from the live API.')
except Exception as e:
    print('  Network unavailable (', type(e).__name__, ') - using inline sample.')
    posts_data = [
        {'userId': 1, 'id': 1, 'title': 'sunt aut facere', 'body': 'quia et suscipit'},
        {'userId': 1, 'id': 2, 'title': 'qui est esse', 'body': 'est rerum tempore vitae'},
        {'userId': 1, 'id': 3, 'title': 'ea molestias quasi', 'body': 'et iusto sed quo iure'}
    ]

# 2. build the DataFrame
posts_df = pd.json_normalize(posts_data)

# 3. shape + first 3 rows
print('\nLab Ex 5 Results:')
print('  Posts DataFrame Shape:', posts_df.shape)
print('  First 3 rows:')
print(posts_df.head(3))
```
