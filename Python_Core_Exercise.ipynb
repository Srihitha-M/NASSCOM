```python
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")

# --- Global Data for the Exercise (Simulated Data) ---
DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0), # Celsius
    'Humidity_Sensor': (30.0, 80.0),   # Percentage
    'Pressure_Sensor': (980.0, 1030.0),# hPa
    'Light_Sensor': (100, 1000)        # Lux
}

# --- Task 1: Python Fundamentals, Lists, Tuples, Sets, Dictionaries, Indexing & Slicing ---
print("\n--- Task 1: Data Generation and Basic Structures ---")

# 1.1 Generate a list of 100 simulated sensor readings.
raw_sensor_data = []
start_time = datetime.datetime(2026, 1, 1, 8, 0, 0)

for i in range(1, 101):
    timestamp = start_time + datetime.timedelta(minutes=15 * i)
    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')
    device_id = f"DEV{i:03d}"
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)
    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    raw_sensor_data.append((timestamp_str, device_id, device_type, location, value))

print(f"Generated {len(raw_sensor_data)} raw sensor readings.")

# 1.2 Temperature readings filter
temperature_readings = [
    {
        'timestamp': item[0],
        'device_id': item[1],
        'location': item[3],
        'temperature_celsius': item[4]
    }
    for item in raw_sensor_data
    if item[2] == 'Temperature_Sensor'
]
print(f"Extracted {len(temperature_readings)} temperature readings.")

# 1.3 Unique locations set
unique_locations = {reading[3] for reading in raw_sensor_data}
print(f"Unique locations found: {unique_locations}")

# 1.4 Calculate device statistics
device_stats = {}
for dtype in DEVICE_TYPES:
    values = [item[4] for item in raw_sensor_data if item[2] == dtype]
    if values:
        device_stats[dtype] = (min(values), max(values))
    else:
        device_stats[dtype] = (None, None)

print("Device Stats (Min, Max):", device_stats)

# 1.5 Advanced Indexing and Slicing
print("\n--- Task 1.5: Advanced Indexing and Slicing (Lists and Strings) ---")
sample_string = "PythonProgrammingIsFun"

# 1.5.1 List Slicing: 10th to 20th elements (indices 9 through 19 inclusive)
sliced_readings_1 = raw_sensor_data[9:20]
# Every 5th reading starting from index 0
sliced_readings_2 = raw_sensor_data[::5]

# 1.5.2 String Slicing: Extract "Programming" and reverse of "Python"
substring_1 = sample_string[6:17]   # "Programming"
substring_2 = sample_string[5::-1]  # "nohtyP"

# 1.5.3 Mixed Slicing
fifth_reading_device_type = raw_sensor_data[4][2]
device_type_suffix = fifth_reading_device_type[-6:]

print(f"10th-20th slice count: {len(sliced_readings_1)} | Every 5th count: {len(sliced_readings_2)}")
print(f"Substring 1: '{substring_1}' | Substring 2: '{substring_2}'")
print(f"5th reading device type: '{fifth_reading_device_type}' | Suffix: '{device_type_suffix}'")


# --- Task 2: Writing and Using Functions in Python ---
print("\n--- Task 2: Function Definitions and Usage ---")

# 2.1 Function: get_readings_by_location
def get_readings_by_location(readings_list, target_location):
    """Filters and returns readings for a specific location."""
    return [reading for reading in readings_list if reading[3] == target_location]

# 2.2 Function: calculate_average_temperature
def calculate_average_temperature(temp_data_list):
    """Calculates average temperature from a list of temperature dictionaries."""
    if not temp_data_list:
        return 0.0
    total_temp = sum(item['temperature_celsius'] for item in temp_data_list)
    return round(total_temp / len(temp_data_list), 2)

# 2.3 Function: generate_report_line
def generate_report_line(reading_dict):
    """Formats a temperature reading dictionary into a human-readable string line."""
    return (
        f"[{reading_dict['timestamp']}] Device {reading_dict['device_id']} "
        f"at {reading_dict['location']}: {reading_dict['temperature_celsius']}°C"
    )

# Execution test for Task 2
lab1_readings = get_readings_by_location(raw_sensor_data, 'Lab1')
avg_temp = calculate_average_temperature(temperature_readings)
print(f"Readings in Lab1: {len(lab1_readings)}")
print(f"Average temperature across all sensors: {avg_temp}°C")


# --- Task 3: Working with Core Python Libraries (os, datetime, random) ---
print("\n--- Task 3: Core Library Usage ---")

# 3.1 Directory Creation
reports_dir = 'sensor_reports'
if not os.path.exists(reports_dir):
    os.makedirs(reports_dir)
    print(f"Directory '{reports_dir}' created.")
else:
    print(f"Directory '{reports_dir}' already exists.")

# 3.2 Time Duration Calculation
if len(raw_sensor_data) > 1:
    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]

    date_format = '%Y-%m-%d %H:%M:%S'
    first_dt = datetime.datetime.strptime(first_timestamp_str, date_format)
    last_dt = datetime.datetime.strptime(last_timestamp_str, date_format)

    duration = last_dt - first_dt

    days = duration.days
    seconds = duration.total_seconds()
    hours = int(seconds // 3600) % 24
    minutes = int((seconds % 3600) // 60)

    print(f"Duration of data collection: {days} days, {hours} hours, {minutes} minutes.")
else:
    print("Not enough data to calculate duration.")

# 3.3 Random Integers
random_values = [random.randint(100, 200) for _ in range(5)]
print(f"Generated random integers: {random_values}")


# --- Task 4: File Handling in Python ---
print("\n--- Task 4: File I/O ---")

# 4.1 Write Report File
report_filename = os.path.join(reports_dir, 'temperature_report.txt')
try:
    with open(report_filename, 'w', encoding='utf-8') as f:
        for reading_dict in temperature_readings:
            line = generate_report_line(reading_dict)
            f.write(line + '\n')
    print(f"Temperature report written to '{report_filename}'.")
except IOError as e:
    print(f"Error writing file '{report_filename}': {e}")

# 4.2 Read Report File Back
read_report_lines = []
try:
    with open(report_filename, 'r', encoding='utf-8') as f:
        read_report_lines = [line.strip() for line in f.readlines()]

    print("\nFirst 3 lines of the report:")
    for line in read_report_lines[:3]:
        print("  ", line)
except IOError as e:
    print(f"Error reading file '{report_filename}': {e}")

print("\n--- Exercise Complete! ---")
print("Great job completing the intermediate Python exercise.")
print("Review your code and outputs to ensure correctness and understanding of all concepts.")
```
