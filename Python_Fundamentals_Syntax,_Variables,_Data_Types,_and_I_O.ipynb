```python
# Python Syntax
print("Hello, Python!")

if True:
    print("This is indented correctly.")
else:
    print("This won't run.")
```

```python
# Variables
name = "Alice"
age = 25
height = 5.6

print("Name:", name)
print("Age:", age)
print("Height:", height)

# Dynamic typing
name = 123
print("New value of name:", name)
```

```python
# Integer
a = 10
print(type(a))
```

```python
# Float
b = 3.14
print(type(b))
```

```python
# String
c = "Python"
print(type(c))
# Boolean
d = True
print(type(d))
# List
e = [1, 2, 3, "four"]
print(type(e), e)
# Dictionary
f = {"name": "Alice", "age": 25}
print(type(f), f)
```

```python
# Output
print("Welcome to Python I/O!")
# Input from user
user_name = input("Enter your name: ")
print("Hello,", user_name)
# Type casting input
user_age = int(input("Enter your age: "))
print("You will be", user_age + 1, "next year.")
```
