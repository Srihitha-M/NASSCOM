```python
import numpy as np
import sympy as sp

x = sp.symbols("x")


def g(x_val):
    return x_val**3 + 2 * x_val


# 1. Numerical derivative at x = 2 (use h = 1e-6)
h = 1e-6
num_deriv = (g(2 + h) - g(2)) / h
print("Numerical g'(2):", round(num_deriv, 4))

# 2. Symbolic derivative of x**3 + 2*x
expr = x**3 + 2 * x
sym_deriv = sp.diff(expr, x)
print("Symbolic derivative g'(x):", sym_deriv)

# 3. Evaluate the symbolic derivative at x = 2 and compare
eval_sym_deriv = sym_deriv.subs(x, 2)
print("Symbolic g'(2):", eval_sym_deriv)
print("Match?", np.isclose(float(num_deriv), float(eval_sym_deriv)))
```

```python
import sympy as sp

x, y = sp.symbols("x y")

h2 = x**2 * y + sp.sin(y)

# 1. dh/dx and dh/dy
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

# 2. Assemble the gradient list
grad = [dh_dx, dh_dy]
print("Gradient vector:", grad)

# 3. Evaluate at (x=2, y=0)
grad_at_pt = [g.subs({x: 2, y: 0}) for g in grad]
print("Gradient at (2, 0):", grad_at_pt)
```

```python
import sympy as sp

x = sp.symbols("x")

# y = exp(x**2 + 1); inner = x**2 + 1, inner' = 2x, outer' = exp(inner)

# 1. By hand (as a SymPy expression)
by_hand = sp.exp(x**2 + 1) * (2 * x)

# 2. With sp.diff
by_sympy = sp.diff(sp.exp(x**2 + 1), x)

# 3. Confirm they match
print("By hand :", by_hand)
print("By SymPy:", by_sympy)
print("Match?  ", sp.simplify(by_hand - by_sympy) == 0)
```

```python
import numpy as np

np.random.seed(42)

Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1  # input -> hidden
Wb = np.random.randn(8, 1) * 0.1  # hidden -> output

# 1. Forward pass: z1, h = ReLU(z1), y_hat, loss
z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb
loss = ((y_hat - Yb) ** 2).mean()

# 2. Backward pass: dy, dWb, dh, dz1, dWa
dy = 2 * (y_hat - Yb) / Yb.size
dWb = h.T @ dy
dh = dy @ Wb.T
dz1 = dh * (z1 > 0)
dWa = Xb.T @ dz1

# 3. One step with lr = 0.05; print loss before and after
lr = 0.05
Wa -= lr * dWa
Wb -= lr * dWb

h_new = np.maximum(0, Xb @ Wa)
y_hat_new = h_new @ Wb
loss_new = ((y_hat_new - Yb) ** 2).mean()

print("Loss before:", round(loss, 4))
print("Loss after :", round(loss_new, 4))
```
